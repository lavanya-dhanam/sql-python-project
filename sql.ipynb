```python
#chart 1-barchar
query1="select crop_name,production_kg,season from crop_production"
df = pd.read_sql(query1,con)
plt.figure(figsize=(7,6))
sns.barplot(data=df,x="crop_name",y="production_kg",hue="season")
plt.xlabel("Crop Name")
plt.ylabel("Production (kg)")
plt.title("Crop-wise Production by Season")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#chart 2-stripplot
query2="select location,production_kg,cultivated_area_acres from crop_production"
df1=pd.read_sql(query2,con)
plt.figure(figsize=(10,6))
sns.stripplot(data=df1,x="location",y="production_kg",hue="cultivated_area_acres")
plt.title("Production by Location and Cultivated Area")
plt.xlabel("Location")
plt.ylabel("Production (kg)")
plt.tight_layout()
plt.show()
```

```python
#chart 3-boxplot
query3="select buyer_type,market_price_per_kg from market_sales";
plt.figure(figsize=(10,6))
df=pd.read_sql(query3,con)
sns.boxplot(x="buyer_type",y="market_price_per_kg",data=df,color="lightgreen")
plt.title("Market Price per Kg by Buyer Type")
plt.xlabel("Buyer Type")
plt.ylabel("Market Price per Kg")
plt.show()
```

```python
#chart4-lineplot 
query4="select sale_date,sum(revenue) as total_revenue from market_sales group by sale_date order by sale_date"
plt.figure(figsize=(10,6))
df = pd.read_sql(query4, con)
sns.lineplot(x="sale_date",y="total_revenue",data=df,color="brown",marker="o")
plt.title("Daily Revenue Trend")
plt.xlabel("Sale Date")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)
plt.show()
```

```python
#chart5-piechart
query5="""select buyer_type,sum(revenue) as total_revenue from market_sales
          group by buyer_type"""
plt.figure(figsize=(10,6))
df= pd.read_sql(query5,con)
plt.pie(
    df["total_revenue"],
    labels=df["buyer_type"],
    autopct="%1.1f%%"
)
plt.title("Revenue Distribution by Buyer Type")
plt.legend()
plt.show()
```

```python
#chart6-swarmplot
query6="""select c.crop_name,m.revenue,c.season from market_sales m
            join crop_production c
            on m.crop_id=c.crop_id"""
plt.figure(figsize=(10,6))
df=pd.read_sql(query6, con)
sns.swarmplot(data=df,x="crop_name",y="revenue",hue="season")
plt.title("Revenue Distribution by Crop")
plt.xlabel("Crop Name")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```

```python
query7="""select c.location,m.market_price_per_kg from market_sales m
            join crop_production c 
            on m.crop_id=c.crop_id"""
plt.figure(figsize=(10,6))
df=pd.read_sql(query7,con)
sns.violinplot(data=df,x="location",y="market_price_per_kg",color="purple",alpha=0.5)
plt.title("Market Price per Kg by Location")
plt.xlabel("Location")
plt.ylabel("Market Price per Kg")
plt.xticks(rotation=45)
plt.show()
```

```python
#chart8-scatterplot
query8="""select c.crop_name,m.quantity_sold_kg,m.market_price_per_kg from market_sales m
          join crop_production c
          on m.crop_id=c.crop_id"""
plt.figure(figsize=(10,6))
df=pd.read_sql(query8,con)
sns.scatterplot(data=df,x="quantity_sold_kg",y="market_price_per_kg",hue="crop_name")
plt.title("Quantity Sold vs Market Price")
plt.xlabel("Quantity Sold (kg)")
plt.ylabel("Market Price per Kg")
plt.show()
```

```python
#chart9-histplot
query9="""select c.crop_name, m.revenue from market_sales m
             join crop_production c
             on m.crop_id=c.crop_id
            where m.revenue>(select avg(revenue)
                             from market_sales) """
plt.figure(figsize=(10,6))
df = pd.read_sql(query9,con)
sns.histplot(data=df,x="revenue",hue="crop_name",kde=True)
plt.title("Revenue Distribution Above Average")
plt.xlabel("Revenue")
plt.ylabel("Count")
plt.show()
```

```python
#chart10-pairplot
query10="""select c.crop_name,c.season,m.quantity_sold_kg,m.market_price_per_kg,m.revenue
           from market_sales m
           join crop_production c
           on m.crop_id=c.crop_id
           where m.revenue>(select avg(revenue)
                            from market_sales)"""
plt.figure(figsize=(10,6))
df=pd.read_sql(query10,con)
sns.pairplot(df,vars=["quantity_sold_kg","market_price_per_kg","revenue"],hue="season")
plt.show()
```

```python
fig,axes=plt.subplots(2, 2, figsize=(15, 10))
#1.Histogram
query1="select production_kg from crop_production"
df1=pd.read_sql(query1,con)
sns.histplot(data=df1,x="production_kg",bins=10,kde=True,ax=axes[0,0],color="green")
axes[0,0].set_title("Production Distribution")
axes[0,0].set_xlabel("Production (Kg)")

#2.Violin Plot
query2="select buyer_type,market_price_per_kg from market_sales"
df2=pd.read_sql(query2,con)
sns.violinplot(data=df2,x="buyer_type",y="market_price_per_kg",ax=axes[0,1],color="brown")
axes[0,1].set_xlabel("buyer_type")
axes[0,1].set_title("Market Price Distribution")
axes[0,1].set_ylabel("Market Price per Kg")

#3.Area Chart
query3="select sale_date, revenue from market_sales order by sale_date"
df3=pd.read_sql(query3, con)
df3["sale_date"]=pd.to_datetime(df3["sale_date"])
axes[1,0].fill_between(
    df3["sale_date"],
    df3["revenue"]
)
axes[1,0].set_title("Revenue Trend")
axes[1,0].set_xlabel("Sale Date")
axes[1,0].set_ylabel("Revenue")
axes[1,0].tick_params(axis="x", rotation=45)


#4.Heatmap
query4="select quantity_sold_kg,market_price_per_kg,revenue from market_sales"
df4=pd.read_sql(query4, con)
sns.heatmap(df4.corr(),annot=True,ax=axes[1,1])
axes[1,1].set_title("Sales Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2,3, figsize=(18, 10))


# 1. DONUT CHART
query1 = """
SELECT m.buyer_type,
       SUM(m.revenue) AS total_revenue
FROM market_sales m
JOIN crop_production c
ON m.crop_id = c.crop_id
GROUP BY m.buyer_type
"""

df1 = pd.read_sql(query1, con)

axes[0,0].pie(
    df1["total_revenue"],
    labels=df1["buyer_type"],
    autopct="%1.1f%%",
    startangle=90,
    wedgeprops=dict(width=0.4)
)

axes[0,0].set_title("Revenue by Buyer Type")


# 2. LINE CHART
query2 = """
SELECT m.sale_date,
       SUM(m.quantity_sold_kg) AS total_quantity
FROM market_sales m
JOIN crop_production c
ON m.crop_id = c.crop_id
GROUP BY m.sale_date
ORDER BY m.sale_date
"""

df2 = pd.read_sql(query2, con)

df2["sale_date"] = pd.to_datetime(df2["sale_date"])

axes[0,1].plot(
    df2["sale_date"],
    df2["total_quantity"],
    marker="o"
)

axes[0,1].set_title("Quantity Sold Over Time")
axes[0,1].set_xlabel("Sale Date")
axes[0,1].set_ylabel("Quantity Sold (Kg)")
axes[0,1].tick_params(axis="x", rotation=45)


# 3. STEP PLOT
query3 = """
SELECT c.season,
       SUM(c.production_kg) AS total_production
FROM crop_production c
JOIN market_sales m
ON c.crop_id = m.crop_id
GROUP BY c.season
"""

df3 = pd.read_sql(query3, con)

axes[0,2].step(
    df3["season"],
    df3["total_production"],
    where="mid"
)

axes[0,2].set_title("Production by Season")
axes[0,2].set_xlabel("Season")
axes[0,2].set_ylabel("Production (Kg)")


# 4. LOLLIPOP CHART
query4 = """
SELECT c.location,
       SUM(m.revenue) AS total_revenue
FROM crop_production c
JOIN market_sales m
ON c.crop_id = m.crop_id
GROUP BY c.location
"""

df4 = pd.read_sql(query4, con)

x = range(len(df4))

axes[1,0].stem(
    x,
    df4["total_revenue"]
)

axes[1,0].set_xticks(x)
axes[1,0].set_xticklabels(
    df4["location"],
    rotation=45
)

axes[1,0].set_title("Revenue by Location")
axes[1,0].set_xlabel("Location")
axes[1,0].set_ylabel("Revenue")


# 5. BAR CHART
query5 = """
SELECT c.crop_name,
       AVG(m.market_price_per_kg) AS avg_price
FROM crop_production c
JOIN market_sales m
ON c.crop_id = m.crop_id
GROUP BY c.crop_name
"""

df5 = pd.read_sql(query5, con)

axes[1,1].bar(
    df5["crop_name"],
    df5["avg_price"]
)

axes[1,1].set_title("Average Market Price by Crop")
axes[1,1].set_xlabel("Crop Name")
axes[1,1].set_ylabel("Average Price per Kg")
axes[1,1].tick_params(axis="x", rotation=45)


# 6. BOX PLOT
query6 = """
SELECT m.buyer_type,
       m.market_price_per_kg
FROM market_sales m
JOIN crop_production c
ON m.crop_id = c.crop_id
"""

df6 = pd.read_sql(query6, con)

sns.boxenplot(
    data=df6,
    x="buyer_type",
    y="market_price_per_kg",
    ax=axes[1,2]
)

axes[1,2].set_title("Market Price by Buyer Type")
axes[1,2].set_xlabel("Buyer Type")
axes[1,2].set_ylabel("Market Price per Kg")


plt.tight_layout()
plt.show()
```
